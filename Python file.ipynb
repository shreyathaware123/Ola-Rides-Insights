## Step 2: Data Cleaning & Preprocessing

```python
#  Convert Date column

df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Date']
```

```python
# Check for duplicates

df.duplicated().sum()
```

```python
# Convert Time column to datetime
df['Time'] = pd.to_datetime(df['Time'], format='mixed').dt.time
```

```python
df['Time']
```

```python
df['Hour'] = df['Time'].apply(lambda x: x.hour)
df['Hour']
```

```python
# Creating a new column 'DayOfWeek' from the 'Date' column

df['DayOfWeek'] = df['Date'].dt.day_name()

# Check the first 5 rows
df[['Date', 'DayOfWeek']].head()
```

### Feature Engineering / Derived Columns

```python
# Creating a new column 'Month' from the 'Date' column

df['Month'] = df['Date'].dt.month

# Check the first 5 rows
df[['Date', 'Month']].head()
```

```python
### Completion ###

# Create IsCompleted column: 1 if ride was successful, 0 otherwise
df['IsCompleted'] = df['Booking_Status'].apply(lambda x: 1 if x == 'Success' else 0)

# Check first 5 rows
df[['Booking_Status', 'IsCompleted']].head()
```

```python
### Cancellation Type (Customer vs Driver) ###


# Create a new column 'Cancellation_Type'
def get_cancellation_type(row):
    if row['Booking_Status'] == 'Canceled by Customer':
        return 'Customer'
    elif row['Booking_Status'] == 'Canceled by Driver':
        return 'Driver'
    else:
        return 'Success'

df['Cancellation_Type'] = df.apply(get_cancellation_type, axis=1)

# Check first 5 rows
df[['Booking_Status', 'Cancellation_Type']].head()
```

# Replacing nulls with replacement on Categorical Columns

```python
# Verify columns exist

df[
    [
        'Canceled_Rides_by_Customer',
        'Canceled_Rides_by_Driver',
        'Incomplete_Rides_Reason',
        'Payment_Method'
    ]
].isnull().sum()
```

```python
# Replace NULLs

df['Canceled_Rides_by_Customer'] = df['Canceled_Rides_by_Customer'].fillna('Not Cancelled')

df['Canceled_Rides_by_Driver'] = df['Canceled_Rides_by_Driver'].fillna('Not Cancelled')

df['Incomplete_Rides_Reason'] = df['Incomplete_Rides_Reason'].fillna('No Failure')

df['Payment_Method'] = df['Payment_Method'].fillna('No Transaction')
```

```python
# Verify replacements

df[
    [
        'Canceled_Rides_by_Customer',
        'Canceled_Rides_by_Driver',
        'Incomplete_Rides_Reason',
        'Payment_Method'
    ]
].isnull().sum()
```

```python
df['Incomplete_Rides'] = df['Incomplete_Rides'].fillna('Not Applicable')
df['Incomplete_Rides']
```
